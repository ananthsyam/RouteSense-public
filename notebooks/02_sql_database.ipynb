```python
from pathlib import Path
import pandas as pd

ROOT = Path(r"<PROJECT_ROOT>")
DATA = ROOT / "data"


files = [
    DATA / "cleaned" / "routesense_orders.csv",
    DATA / "synthetic" / "routesense_orders_enriched.csv",
    DATA / "synthetic" / "drivers.csv",
    DATA / "synthetic" / "vehicles.csv",
    DATA / "synthetic" / "routes.csv",
    DATA / "synthetic" / "delivery_attempts.csv"
]

for file in files:
    print(f"{file.name:40} -> {file.exists()}")

orders = pd.read_csv(
    DATA / "synthetic" / "routesense_orders_enriched.csv"
)

drivers = pd.read_csv(
    DATA / "synthetic" / "drivers.csv"
)

vehicles = pd.read_csv(
    DATA / "synthetic" / "vehicles.csv"
)

routes = pd.read_csv(
    DATA / "synthetic" / "routes.csv"
)

delivery_attempts = pd.read_csv(
    DATA / "synthetic" / "delivery_attempts.csv"
)


print("Orders           :", orders.shape)
print("Drivers          :", drivers.shape)
print("Vehicles         :", vehicles.shape)
print("Routes           :", routes.shape)
print("Delivery attempts:", delivery_attempts.shape)
```

```python
import os
import mysql.connector
from mysql.connector import Error

password = os.getenv("ROUTESENSE_MYSQL_PASSWORD")

try:
    conn = mysql.connector.connect(
        host="localhost",
        user="root",
        password=password
    )

    if conn.is_connected():
        print("MySQL connection successful")
        print("Server version:", conn.get_server_info())

except Error as e:
    print("MySQL connection failed")
    print("Error:", e)
```

```python
cursor = conn.cursor()

cursor.execute("CREATE DATABASE IF NOT EXISTS routesense")
cursor.execute("USE routesense")

print("Database selected:", conn.database)
```
